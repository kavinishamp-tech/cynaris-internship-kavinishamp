```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("NumPy Version:", np.__version__)
print("Pandas Version:", pd.__version__)
```

```python
arr_1d = np.array([10, 20, 30, 40])

arr_2d = np.array([
    [10, 20],
    [30, 40]
])

arr_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D Shape:", arr_1d.shape)
print("2D Shape:", arr_2d.shape)
print("3D Shape:", arr_3d.shape)
```

```python
# Broadcasting

marks = np.array([70, 80, 90])

bonus = 5

updated_marks = marks + bonus

print("Original Marks:", marks)
print("Bonus:", bonus)
print("Updated Marks:", updated_marks)
```

```python
# Vectorized Operations

numbers = np.array([1, 2, 3, 4, 5])

doubled = numbers * 2
squared = numbers ** 2

print("Original:", numbers)
print("Doubled:", doubled)
print("Squared:", squared)
```

```python
# Matrix Multiplication

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

result = A @ B

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\nA @ B:")
print(result)
```

```python
import pandas as pd
from pathlib import Path

# Find project directory
project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

# Reload CSV dataset
df = pd.read_csv(project_root / "data" / "student_scores.csv")

# Remove accidental spaces from column names
df.columns = df.columns.str.strip()

# Display column names
print("Columns:", df.columns.tolist())

# Display dataset
display(df)
```

```python
scores = df[["Math", "Science", "English"]].to_numpy()

print("NumPy Scores:")
print(scores)

print("Shape:", scores.shape)
```

```python
subject_mean = np.mean(scores, axis=0)

print("Mean Marks")
print("Math:", subject_mean[0])
print("Science:", subject_mean[1])
print("English:", subject_mean[2])
```

```python
subject_std = np.std(scores, axis=0)

print("Standard Deviation")
print("Math:", subject_std[0])
print("Science:", subject_std[1])
print("English:", subject_std[2])
```

```python
correlation = np.corrcoef(scores, rowvar=False)

print("Correlation Matrix:")
print(correlation)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(["Math", "Science", "English"], subject_mean)

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.show()
```
